```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
# Load Dataset

df = pd.read_csv('/content/Sample - Superstore.csv', encoding='latin1')

df.head()
```

```python
# Dataset Overview

df.shape
df.columns
df.info()
df.describe()
```

```python
# Missing Values

df.isnull().sum()
```

```python
# Total Sales

total_sales = df['Sales'].sum()
print(total_sales)
```

```python
# Total Profit

total_profit = df['Profit'].sum()
print(total_profit)
```

```python
# Sales by Category

sales_by_category = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

sales_by_category
```

```python
sales_by_category.plot(kind='bar')

plt.title('Sales by Category')
plt.ylabel('Sales')
plt.show()
```

```python
# Profit by Category

profit_by_category = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

profit_by_category
```

```python
profit_by_category.plot(kind='bar')

plt.title('Profit by Category')
plt.ylabel('Profit')

plt.show()
```

```python
# Top 10 Products by Sales

top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_products
```

```python
top_products.plot(kind='bar')

plt.title('Top 10 Products by Sales')

plt.show()
```

```python
# Sales by Region

region_sales = df.groupby('Region')['Sales'].sum()

region_sales.plot(kind='pie', autopct='%1.1f%%')

plt.ylabel('')

plt.show()
```

Business Insights

1. Technology generated the highest sales.

2. Some categories generated strong revenue but lower profitability.

3. Certain products contributed significantly to overall sales.

4. Regional performance varied across the business.

Recommendations

- Focus marketing efforts on high-performing categories.

- Review low-profit products.
